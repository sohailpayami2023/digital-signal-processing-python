# 1. Variables and Types

Python variables are **dynamically typed** — you do not
declare a type, Python infers it from the value.
MATLAB works the same way, but Python has more distinct
types that matter in practice.

| Type | Example | MATLAB equivalent |
|------|---------|-------------------|
| `int` | `n = 10` | `n = 10` |
| `float` | `x = 3.14` | `x = 3.14` |
| `complex` | `z = 1 + 2j` | `z = 1 + 2i` |
| `bool` | `flag = True` | `flag = true` |
| `str` | `s = "hello"` | `s = "hello"` |
| `None` | `val = None` | `val = []` (loosely) |


````python
# Integer — MATLAB: n = 10
n = 10
print(type(n), n)   # <class 'int'> 10

# Float — MATLAB: x = 3.14
x = 3.14
print(type(x), x)   # <class 'float'> 3.14

# Complex — Python uses j, MATLAB uses i or j
z = 1 + 2j
print(type(z), z)   # <class 'complex'> (1+2j)
print('real:', z.real, ' imag:', z.imag)

# Boolean — Python uses True/False (capital T/F)
# MATLAB uses true/false (lowercase)
flag = True
print(type(flag), flag)   # <class 'bool'> True

# None — Python equivalent of null/empty
# Often used as a default function argument
result = None
print(result is None)   # True

# Type casting
# int() / float() — MATLAB: int32(x) / double(x)
print(int(3.9))    # 3  (truncates, does not round)
print(float(5))    # 5.0
print(str(42))     # '42'
````

# 2. `if` / `elif` / `else`

Python uses **indentation** (4 spaces) to define blocks —
there is no `end` keyword like in MATLAB.

```
MATLAB                  Python
----------------------  ----------------------
if x > 0               if x > 0:
    disp("pos")            print("pos")
elseif x == 0          elif x == 0:
    disp("zero")           print("zero")
else                   else:
    disp("neg")            print("neg")
end                    # (no end needed)
```

Comparison operators are the same: `>`, `<`, `>=`, `<=`,
`==`, `!=`.
Logical operators: `and`, `or`, `not`
(MATLAB: `&&`, `||`, `~` or `!`).


````python
snr_db = 15

# if / elif / else — MATLAB: if / elseif / else / end
if snr_db >= 20:
    regime = "high SNR"
elif snr_db >= 10:
    regime = "medium SNR"
else:
    regime = "low SNR"

print(regime)   # medium SNR

# Logical operators: and, or, not
# MATLAB: &&, ||, ~
modulation = "QPSK"
if snr_db > 5 and modulation == "QPSK":
    print("QPSK viable at this SNR")

# Membership test — MATLAB: ismember(x, list)
if modulation in ['BPSK', 'QPSK', '16-QAM', '64-QAM']:
    print("Valid modulation scheme")

# Inline (ternary) if — MATLAB: no direct equivalent
# value_if_true if condition else value_if_false
label = "pass" if snr_db > 10 else "fail"
print(label)   # pass
````

# 3. `for` and `while` Loops

```
MATLAB                      Python
------------------------    -------------------------
for i = 1:10               for i in range(10):
    disp(i)                    print(i)
end

for i = 1:2:10             for i in range(1, 11, 2):
    disp(i)                    print(i)
end
```

`range(start, stop, step)` — `stop` is **exclusive**.
Looping directly over a list is preferred over index loops.


````python
# Basic for loop — MATLAB: for i = 1:5 ... end
# range(5) gives 0, 1, 2, 3, 4
for i in range(5):
    print(i, end=" ")
print()   # newline

# range(start, stop, step) — MATLAB: 0:2:8
for i in range(0, 10, 2):
    print(i, end=" ")
print()

# Loop directly over a list (no index needed)
# MATLAB: for mod = {"BPSK","QPSK","16-QAM"} ... end
schemes = ['BPSK', 'QPSK', '16-QAM']
for scheme in schemes:
    print('Testing:', scheme)

# Accumulate results into a list
# MATLAB: results = zeros(1,5); for i=1:5 ...
results = []
for i in range(5):
    results.append(i ** 2)
print('squares:', results)
````

````python
import numpy as np

# Typical DSP loop: sweep SNR, compute BER
snr_range = range(0, 25, 5)   # 0, 5, 10, 15, 20 dB

ber_values = []
for snr in snr_range:
    # Theoretical BPSK BER: 0.5 * erfc(sqrt(10^(SNR/10)))
    snr_lin = 10 ** (snr / 10)
    ber = 0.5 * np.exp(-snr_lin)   # simplified
    ber_values.append(ber)

print('SNR (dB) | BER')
for snr, ber in zip(snr_range, ber_values):
    print(f'  {snr:5} dB | {ber:.2e}')

# while loop — MATLAB: while condition ... end
count = 0
while count < 3:
    print('count =', count)
    count += 1   # MATLAB: count = count + 1
````

# 4. Functions

```
MATLAB                       Python
---------------------------  ---------------------------
function y = f(x)           def f(x):
    y = x^2;                    return x**2
end

function [a,b] = g(x)       def g(x):
    a = x+1; b = x-1;           return x+1, x-1
end
```

Key differences from MATLAB:
- `def` instead of `function`
- No `end`
- Return multiple values as a tuple (automatic unpacking)
- Default argument values are set in the signature
- Arguments can be passed by name (keyword arguments)


````python
# Basic function — MATLAB: function y = square(x) ... end
def square(x):
    return x ** 2

print(square(5))    # 25

# Default arguments — MATLAB: no direct syntax
# If fs is not provided, it defaults to 1000
def make_sine(freq, duration, fs=1000):
    t = np.linspace(0, duration, int(fs * duration),
                    endpoint=False)
    return t, np.sin(2 * np.pi * freq * t)

# Call with defaults
t1, s1 = make_sine(50, 0.1)
print('default fs, samples:', len(s1))   # 100

# Override with keyword argument
t2, s2 = make_sine(50, 0.1, fs=8000)
print('fs=8000,  samples:', len(s2))   # 800
````

````python
# Return multiple values — MATLAB: [a, b] = f(x)
def stats(signal):
    """Return mean, std, and peak of a signal array."""
    return np.mean(signal), np.std(signal), np.max(
        np.abs(signal)
    )

sig = np.random.randn(1000)
# Unpack all three return values at once
mean, std, peak = stats(sig)
print(f'mean={mean:.3f}  std={std:.3f}  peak={peak:.3f}')

# A realistic DSP helper: add AWGN to a signal
def add_awgn(signal, snr_db):
    """Add white Gaussian noise at the given SNR (dB).

    Parameters
    ----------
    signal : np.ndarray  Clean input signal
    snr_db : float       Desired SNR in dB

    Returns
    -------
    np.ndarray  Noisy signal
    """
    # Signal power (mean square)
    sig_power = np.mean(np.abs(signal) ** 2)
    # Noise power from SNR: SNR = P_signal / P_noise
    noise_power = sig_power / (10 ** (snr_db / 10))
    noise = np.sqrt(noise_power) * np.random.randn(
        len(signal)
    )
    return signal + noise

clean = np.sin(2 * np.pi * 0.1 * np.arange(100))
noisy = add_awgn(clean, snr_db=10)
print('Noisy signal generated, length:', len(noisy))
````

# 5. List Comprehensions

A compact way to build a list from a loop — very common
in Python DSP code. MATLAB has no direct equivalent;
closest is `arrayfun`.

```python
# Equivalent to:
result = []
for x in iterable:
    result.append(expression)

# Written as:
result = [expression for x in iterable]

# With a filter:
result = [expression for x in iterable if condition]
```


````python
# Squares of 0–9 — MATLAB: (0:9).^2
squares = [i**2 for i in range(10)]
print('squares:', squares)

# Convert SNR list from dB to linear
# MATLAB: 10.^(snr_db/10)
snr_db_list = [0, 5, 10, 15, 20]
snr_lin = [10**(s/10) for s in snr_db_list]
print('linear SNR:', [round(s, 2) for s in snr_lin])

# Filter: keep only even numbers
evens = [i for i in range(20) if i % 2 == 0]
print('evens:', evens)

# Build a list of label strings
# MATLAB: arrayfun(@(s) sprintf(...), snr_db, ...)
labels = [f"SNR = {s} dB" for s in snr_db_list]
for label in labels:
    print(label)
````

# 6. Lists, Tuples, and Dictionaries

| Type | Syntax | Mutable | MATLAB equivalent |
|------|--------|---------|-------------------|
| List | `[1, 2, 3]` | Yes | Cell array |
| Tuple | `(1, 2, 3)` | No | — |
| Dict | `{"a": 1}` | Yes | struct / containers.Map |

**Lists** are the general-purpose sequence — use them
when you need to grow/modify a collection.

**Tuples** are immutable lists — used for fixed groupings
such as `(rows, cols)` shape or function return values.

**Dictionaries** store named parameters — ideal for
simulation configs (modulation, SNR range, frame size, …).


````python
# List — MATLAB: c = {1, "hello", [1 2 3]}
items = [1, 'hello', [1, 2, 3]]
print('list  :', items)
items.append('new')       # add to end
print('after append:', items)
print('last item   :', items[-1])  # 'new'

# Tuple — immutable, often used for shapes
shape = (3, 4)     # (rows, cols)
print('tuple:', shape)
print('rows :', shape[0])

# Unpack tuple — MATLAB: [a, b] = deal(1, 2)
rows, cols = shape
print(f'{rows} rows, {cols} cols')
````

````python
# Dictionary — MATLAB: params.modulation = "QPSK"
sim_params = {
    'modulation' : 'QPSK',
    'snr_range'  : list(range(0, 25, 5)),
    'n_bits'     : 100_000,
    'fs'         : 1000,
}

# Access by key — MATLAB: params.modulation
print(sim_params['modulation'])    # QPSK

# Add or update a key
sim_params['code_rate'] = 0.5

# Check if key exists — MATLAB: isfield(params, "fs")
print('fs' in sim_params)   # True

# Loop over key-value pairs
# MATLAB: fields = fieldnames(params); for ...
for key, value in sim_params.items():
    print(f'  {key}: {value}')
````

# 8. String Formatting with f-strings

f-strings (Python 3.6+) are the cleanest way to build
output strings. Prefix the string with `f` and embed
expressions inside `{}`.

MATLAB equivalent: `sprintf('SNR = %d dB', snr)`

| Format code | Meaning | Example output |
|-------------|---------|----------------|
| `{x}` | Default | `3.14159...` |
| `{x:.2f}` | 2 decimal places | `3.14` |
| `{x:.4e}` | Scientific notation | `3.1416e+00` |
| `{x:d}` | Integer | `42` |
| `{x:>10}` | Right-align in 10 chars | `      3.14` |
| `{x:.2%}` | Percentage | `31.42%` |


````python
snr = 15
ber = 1.23e-4
scheme = "QPSK"

# Basic f-string — MATLAB: sprintf('%s at SNR=%d', ...)
print(f'{scheme} at SNR = {snr} dB')

# Fixed decimal places
print(f'BER = {ber:.2e}')     # BER = 1.23e-04
print(f'BER = {ber:.6f}')     # BER = 0.000123

# Table formatting — align columns with width specifiers
print(f"{'Scheme':>8} | {'SNR':>6} | {'BER':>10}")
print('-' * 32)
for snr_val in [5, 10, 15, 20]:
    ber_val = 0.5 * 10 ** (-snr_val / 10)
    print(f'{scheme:>8} | {snr_val:>6} | {ber_val:>10.2e}')
````

# 9. Classes

A class bundles data and the functions that operate on it.
In MATLAB you might use a `struct` for data and separate
functions — a Python class keeps them together.

For DSP/wireless work you will mostly **use** classes
(e.g. Scikit-learn estimators, PyTorch `nn.Module`)
rather than write your own. But knowing the basics helps
you read library code and write reusable simulators.

```
Key terms:
  __init__  : constructor  (MATLAB: obj = MyClass(...))
  self      : the object itself (MATLAB: obj)
  attribute : data stored on the object (obj.snr)
  method    : function defined inside the class
```


````python
# A simple AWGN channel class
class AWGNChannel:
    """Models an Additive White Gaussian Noise channel.

    Parameters
    ----------
    snr_db : float  Signal-to-noise ratio in dB
    """

    def __init__(self, snr_db):
        # Store SNR as an attribute on the object
        self.snr_db  = snr_db
        self.snr_lin = 10 ** (snr_db / 10)

    def transmit(self, signal):
        """Add noise and return the noisy signal."""
        sig_power   = np.mean(np.abs(signal) ** 2)
        noise_power = sig_power / self.snr_lin
        noise = np.sqrt(noise_power) * np.random.randn(
            len(signal)
        )
        return signal + noise

    def __repr__(self):
        """String shown when you print the object."""
        return f'AWGNChannel(snr_db={self.snr_db})'


# Create an instance — MATLAB: ch = AWGNChannel(10)
ch = AWGNChannel(snr_db=10)
print(ch)              # AWGNChannel(snr_db=10)
print(ch.snr_lin)      # 10.0

# Call a method
clean = np.sin(2 * np.pi * 0.05 * np.arange(200))
noisy = ch.transmit(clean)
print('signal len:', len(noisy))
````

````python
# Inheritance: RayleighChannel extends AWGNChannel
class RayleighChannel(AWGNChannel):
    """Flat Rayleigh fading channel with AWGN."""

    def transmit(self, signal):
        # Draw complex fading coefficient h ~ CN(0,1)
        h = (np.random.randn() + 1j * np.random.randn())
        h /= np.sqrt(2)   # normalise to unit mean power
        faded = np.abs(h) * signal   # flat fading
        # Call parent class for noise addition
        return super().transmit(faded)


ch_ray = RayleighChannel(snr_db=15)
print(ch_ray)   # AWGNChannel(snr_db=15)
noisy_ray = ch_ray.transmit(clean)
print('Rayleigh+AWGN signal generated')
````

# 11. Error Handling — `try` / `except`

Use `try/except` to handle errors gracefully rather than
crashing. In MATLAB the equivalent is `try/catch`.

```
MATLAB              Python
---------------     ---------------
try                 try:
    ...                 ...
catch e             except ValueError as e:
    disp(e.message)     print(e)
end
```

Common exceptions in NumPy / SciPy work:
- `ValueError` — wrong shape, bad argument value
- `IndexError` — index out of range
- `ImportError` — package not installed
- `FileNotFoundError` — path does not exist


````python
# Basic try/except
def safe_log(x):
    """Return log10(x), or warn if x <= 0."""
    try:
        result = np.log10(x)
        return result
    except Exception as e:
        print(f'Error: {e}')
        return None

print(safe_log(100))    # 2.0
print(safe_log(-1))     # returns nan (not an exception here)

# Raise your own error to enforce constraints
def set_snr(snr_db):
    if snr_db < -10 or snr_db > 50:
        raise ValueError(
            f'SNR {snr_db} dB is outside valid range '
            '(-10 to 50 dB).'
        )
    return snr_db

try:
    set_snr(99)
except ValueError as e:
    print(f'Caught: {e}')
````
